```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('esimple.csv')

```

```python
sns.set(style='white')
```

```python
sns.histplot(df['JoiningYear'],kde=True,bins=10)
plt.title('distribution of age')
```

```python
sns.boxplot(x='JoiningYear',y='Age',hue='JoiningYear',data=df)
plt.title('age distribution')
```

```python
sns.boxplot(x='JoiningYear',y='Age',data=df)
plt.title('age distribution')
```

```python
sns.scatterplot(x='JoiningYear',y='Age',data=df)
plt.title('age distribution')
```
